# Randomly sample 10,000 rows from a CSV file

```python
# Define library
import pandas as pd
```

```python
# Load the CSV file
file = "nlbse23-issue-classification-train.csv"
df = pd.read_csv(file)
```

```python
# Check total number of rows
print("Original dataset size:", len(df))
```

```python
# Random sampling of 10,000 rows
# random_state is used so the sample is reproducible
sample_df = df.sample(n=10000, random_state=42)
print("Sample dataset size:", len(sample_df))
```

```python
# Preview sampled data
print(sample_df.head())
```

```python
# Save sampled dataset to a new CSV file
output_path = "nlbse23-issue-classification-train-sampled10k.csv"
sample_df.to_csv(output_path, index=False)

print(f"10,000 sampled rows saved to: {output_path}")
```
